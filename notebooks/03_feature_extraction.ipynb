# 03 - TicketIQ: Feature Extraction (TF-IDF vs Bag of Words)
Vectorizes preprocessed ticket text into sparse matrices and compares vocabulary sizing and n-gram representations.

```python
import sys
sys.path.append('..')
from src.feature_extraction import get_vectorizer
import pandas as pd

df = pd.read_csv('../data/sample/sample_tickets.csv')
texts = df['Ticket Description'].astype(str).tolist()

tfidf_vec = get_vectorizer(method='tfidf', max_features=5000)
X_tfidf = tfidf_vec.fit_transform(texts)
print('TF-IDF Matrix Shape:', X_tfidf.shape)

bow_vec = get_vectorizer(method='bow', max_features=5000)
X_bow = bow_vec.fit_transform(texts)
print('BoW Matrix Shape:   ', X_bow.shape)
```
