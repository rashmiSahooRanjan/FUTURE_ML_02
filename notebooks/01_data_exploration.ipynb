# 01 - TicketIQ: Exploratory Data Analysis (EDA)
Comprehensive analysis of the customer support ticket dataset, checking distributions, schema, missing values, and data integrity.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/raw/customer_support_tickets.csv')
print('Dataset Shape:', df.shape)
df.head()
```

```python
df.info()
```

```python
print('Missing Values:
', df.isnull().sum())
```

```python
print('Ticket Categories:
', df['Ticket Type'].value_counts())
print('
Ticket Priorities:
', df['Ticket Priority'].value_counts())
```

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=df, x='Ticket Type', palette='Blues_d')
plt.title('Distribution of Support Ticket Categories')
plt.xticks(rotation=15)
plt.show()
```
